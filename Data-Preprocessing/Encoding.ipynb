# Endcoding

Encoding means converting the categorical data into numeric form

## One hot Encoding

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('loan_data_set.csv')
df.head(3)
```

```python
df.isnull().sum()
```

```python
# first filling the missing values
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
```

```python
df.isnull().sum()
```

### Using Get Dummies

```python
en_data = df[['Gender', "Married"]]
en_data.head(3)
```

```python
# pd.get_dummies(en_data)  # returns in true and false
pd.get_dummies(en_data).astype(int)
# pd.get_dummies(en_data).astype(int).info()
```

### Using Scikit Learn

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
# ohe.fit_transform(en_data)    # returns a sparse matrix

ar = ohe.fit_transform(en_data).toarray()
ar
```

```python
en_data.info()
```

```python
pd.DataFrame(ar, columns=["Gender_Female" ,"Gender_Male",	"Married_No",	"Married_Yes"]).head(3)
```

```python
en_data.head()
```

```python
ohe = OneHotEncoder(drop='first')

ar = ohe.fit_transform(en_data).toarray()
ar
```

```python
pd.DataFrame(ar, columns=["Gender_Male", "Married_Yes"]).head(5)
```

## Label Encoding

Used for unsequence/unorderd data

```python
df = pd.DataFrame({
    "name":['cow', 'dog', 'cat', 'white', 'black'],
})
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
le.fit_transform(df['name'])
```

```python
df['en_name'] = le.fit_transform(df['name'])

df
```

## Ordinal Encoding

```python
df = pd.DataFrame({"Size":['s', 'm', 'l', 'xl', 's', 'm', 'l', 's', 's', 'l', 'xl', 'm']})
df.head(3)
```

```python
df['Size'].unique()
```

```python
ord_data = [['s', 'm', 'l', 'xl']]
```

### Using Scikit Learn

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# oe = OrdinalEncoder() # by default it encodes based on alphabets
oe = OrdinalEncoder(categories=ord_data) # here we specify that what should be the order
oe.fit_transform(df[['Size']])
```

```python
df['en_size'] = oe.fit_transform(df[['Size']])
df
```

```python
df['en_size'].unique()
```

### Using Map Function

```python
# create a dictionary first
ord_data1 = {"s":"0", "m":"1", "l":"2", "xl":"3"}
ord_data1
```

```python
df['Size'].map(ord_data1)
```

```python
df['en_size_map'] = df['Size'].map(ord_data1)
df
```

```python
# create a dictionary first
ord_data2 = {"s":"11", "m":"12", "l":"33", "xl":"57"}
ord_data2
```

```python
df['en_size_another'] = df['Size'].map(ord_data2)
```

```python
df
```
