```python
def factorial_recursive(n):
  if n==1 or n==0 :
    return 1
  else:
    return n * factorial_recursive(n-1)

a=int(input("Enter a number: "))
print(factorial_recursive(a))
```

```python
def factorial_iter(n):
  product = 1
  for a in range(n):
    product = product * (a+1)
  return(product)

factorial_iter(5)
```

***Rock Paper Scissor Game***

```python
import random

def game(computer, player) :
  if computer == player :
    return None

  elif computer == 'r':
    if player == 'p':
      return True
    elif player == 's' :
      return False

  elif computer == 'p':
      if player == 's':
        return True
      elif player == 'r' :
        return False

  elif computer == 's':
    if player == 'r':
      return True
    elif player == 'p':
      return False

print("Computer Turn: Rock(r) Paper(p) Scissor(s)")

randomNumber = random.randint(1, 3)  # --> This will give any value fom 1 to 3 including both

if randomNumber == 1:
  computer = 'r'
elif randomNumber == 2:
  computer = 'p'
elif randomNumber == 3:
  computer = 's'

print("Select Option from: (r) (p) (s)")
player = input("Player Turn: Rock(r) Paper(p) Scissor(s): ")

a = game(computer, player)

print(f"Computer chose {computer}")
print(f"Player chose {player}")

if a == None :
  print("Game is a Tie")
elif a :
  print("You Win !")
else :
  print("You Lose !")
```

*Static Method*

A static method is a special type of method in object-oriented programming that belongs to a class rather than an instance of that class. Static methods can be accessed directly from the class without needing to create an object of that class. They are useful when you have code that can be shared across all instances of a class or when you want to perform a calculation or operation without creating an object.

```python
# Instresting one

class Train:
  def __init__(self, name, fare, seats):
    self.name = name
    self.fare = fare
    self.seats = seats

  def getStatus(self):
    print(f"The name of train is {self.name}")
    print(f"The number of seats in train are {self.seats}")

  def fareInfo(self):
    print(f"The price of a one seat is rs. {self.fare}")

  def bookTicket(self):
    if (self.seats>0):
      print(f"Your ticket is booked !!. Your seat no. is {self.seats}")
      self.seats = self.seats - 1
    else:
      print("Seats are full, try to book another train")

a = Train("Maharashtra Express: 11001", 100, 2)
a.getStatus()
a.fareInfo()

a.bookTicket()
a.bookTicket()
a.bookTicket()
```

```python
# Single Inheritance

class Employee:  # --> Base class (Parent class)
  company = "Google"

  def showDetails(self):
    print(f"This is a Employee ")

class Programmer(Employee):  # --> Derived class (Child class)
  language = "Python"
  # company = "Microsoft   # This overwrites the parent class attribute

  def info(self):
    print(f"The Employee do coding in {self.language} ")

  def showDetails(self):
    print(f"This is a programmer")   # This overwrites the parent class attribute

a = Employee()
b = Programmer()

a.showDetails()
b.info()
b.showDetails()

print(b.company)
print(b.language)
```

```python
# Multiple Inheritance

class Employee:
  company = "Microsoft"
  eCode = 123

class FreeLancer:
  company = 'Visa'
  level = 0

  def upgradeLevel(self):
    self.level = self.level + 1

class Coder(Employee, FreeLancer):  # First prefrence is given to Employee class as it is written first
  name = 'Darshan'

e = Employee()
f = FreeLancer()
c = Coder()

c.upgradeLevel()

print(c.name)
print(c.eCode)
print(c.level)
print(c.company)
```

```python
# Multilevel Inheritance

class Person():
  country = "IND"

  def new(self):
    print("I am a Fresher")

class Employee(Person):
  company = "Flipkart"

  def getSalary(self, salary):
    self.salary = salary
    print(f"Salary is {self.salary}")

class Programmer(Employee):
  language = "python"
  # company = "Amazon"   This will over write

  # def getSalary(self):
  #   print("High Salary")

a = Person()
b = Employee()
c = Programmer()

a.new()
c.getSalary(1000)

print(c.country)
print(c.company)
```

```python
class Person():
  country = "IND"

  def new(self):
    print("I am a Fresher")

class Employee(Person):
  company = "Flipkart"

  def new1(self):
    super().new()
    print("I am Employee")

class Programmer(Employee):
  language = "python"

  def new2(self):
    super().new()
    super().new1()
    print("I am Programmer")

a = Person()
b = Employee()
c = Programmer()

a.new()

b.new1()

c.new2()
```

@.setter

5.jpg

```python
class Employee:
  company = "Google"
  salary = 5000
  salarybonus = 1000

  @property
  def totalSalary(self):
    return self.salary + self.salarybonus

  @totalSalary.setter
  def totalSalary(self, val):
    self.salarybonus = val - self.salary

a = Employee()
print(a.totalSalary)

a.totalSalary = 5400
print(a.salary)
print(f"The salary bonus is {a.salarybonus}")
```

Operators Overloading in Python   + Other DUNDER methods

2.png

1.png

```python
class Number:
  def __init__(self, num):
    self.num = num

  def __add__(self, self1):
    print("Adding")
    return self.num + self1.num

  def __sub__(self, self1):
    print("Substracting")
    return self.num - self1.num

  def __mul__(self, self1):
    print("Multiplying")
    return self.num * self1.num

  def __truediv__(self, self1):
    print("Dividing")
    return self.num / self1.num

  def __floordiv__(self, self1):
    print("Floordiv")
    return self.num // self1.num

# OTHER DUNDER METHODS
  def __str__(self):
    return f"The Number is {self.num}"

  def __len__(self):
    return 1

n1 = Number(4)
n2 = Number(5)

sum = n1 + n2
print(sum)

sub = n1 - n2
print(sub)

mul = n1 * n2
print(mul)

div = n1 / n2
print(div)

floordiv = n1 // n2
print(floordiv)


print(n1)
print(n2)

print(len(n1))
print(len(n2))
```

***The Perfect Guess Game***

```python
import random

randomNumber = random.randint(1,100)   # gives any random no. from 1 to 100

guesses = 0
userGuess = None   # this is written to define userGuess

while (userGuess != randomNumber):
    userGuess = int(input("Enter a number to guess: "))
    guesses = guesses + 1
    if (userGuess > randomNumber):
      print("Enter a lower number")
    elif (userGuess < randomNumber):
      print("Enter a higher number")
    else:
      print("Your Guess is correct")

print(f"You guessed correctly in {guesses} guesses")
```

```python
L = [5, 12, 6, 89, 90]

# Print a slice of list L from index 4 to 1. Since the start index is greater than the end index, it won't return any elements.
print(L[4:1])  # Output: []

# Print a slice of list L with a step of -2. This will return every other element in reverse order.
print(L[::-2])  # Output: [90, 6, 5]

# Print a slice of list L from index 1 to 5 with a step of 1. This will return elements from index 1 to 4.
print(L[1:5])  # Output: [12, 6, 89, 90]

s = "REMARKABLE"

# Print the reverse of the string s.
print(s[::-1])  # Output: "ELBAMARKER"

# Print a slice of the string s from index -6 to the end of the string.
print(s[-6:])  # Output: "ARKABLE"

# Print a slice of the string s from index 2 to the second-to-last character with a step of 1.
print(s[2:len(s)-1])  # Output: "MARKABL"
```
