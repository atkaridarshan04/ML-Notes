```python
size = 100000
```

```python
L1 = range(size)
L2 = range(size)

A1 = np.arange(size)
A2 = np.arange(size)
```

```python
start = time.time()
result = [(x+y) for x,y in zip(L1,L2)]
result = [(x+y) for x,y in zip(L1,L2)]

print("Python List Took: ", (time.time() - start) * 1000, "sec")
```

```python
start = time.time()
result = A1 + A2

print("Numpy List Took: ", (time.time() - start) * 1000, "sec")
```

# Adding new Rows and Columns

```python
original = np.array([[1,2], [3,4]])

add_row = np.array([[5,6]])
add_col = np.array([[7],[8]])
```

```python
with_new_row = np.vstack((original, add_row))
print("Added new row: \n", with_new_row)
```

```python
with_new_col = np.hstack((original, add_col))
print("Added new col: \n", with_new_col)
```

# Example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
# Sales data shape
print(sales_data.shape)
```

```python
# Data1st for 1st 3 restaurents
print(sales_data[:3, 1:]) 
```

```python
# total sales for wach restaurent
print(np.sum(sales_data, axis=0))

# total sales per year collectively
print(np.sum(sales_data[:, 1:], axis=1)) 
```

```python
# Minimum slaes per restaurent
print("Minimum", np.min(sales_data[:, 1:], axis=1))

# Maximum slaes per year
print("Maximum ",np.max(sales_data[:, 1:], axis=0))

# Average Sales per restaurent
print("Average ", np.mean(sales_data[:, 1:], axis=1))
# print("Average ", np.average(sales_data[:, 1:], axis=1))
```

```python
# Cumulative Sales for res
cumsum = sales_data[:, 1:]
print("Cumulative Sale\n", np.cumsum((cumsum), axis=1))
```

```python
# graph for cumulative sales
plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))

plt.title("Average cumulative sales accross all restaurant ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
monthly_avg = sales_data[:, 1:] / 12
print(monthly_avg)
```

```python
restaurant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])
vectorized_upper = np.vectorize(str.upper)
print("Vectorized Upper", vectorized_upper(restaurant_types))
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector addition: ", vector1 + vector2)

print("Multiplication vector", vector1 * vector2)

print("Dot Product", np.dot(vector1, vector2))
```

```python
# sin angle betn two lines
# sin(a,b) = (a.b) / (|a|.|b|)

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
