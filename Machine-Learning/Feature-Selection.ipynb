# Feature Selction Techniques

**Forward selection** starts with an empty set of features and gradually adds the most significant ones, while **backward selection** begins with all features and iteratively removes the least significant ones until an optimal feature set is reached.

```python
import pandas as pd
from mlxtend.feature_selection import SequentialFeatureSelector
```

```python
df = pd.read_csv('./datasets/diabetes.csv')
df.head(3)
```

```python
x = df.iloc[:, :-1]
y = df['Outcome']
```

```python
x.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression(max_iter=500)
```

```python
fs = SequentialFeatureSelector(lr, k_features=8, forward=True, cv=None)
fs.fit(x,y)
```

```python
fs.feature_names
```

```python
fs.k_feature_names_
```

```python
fs.k_score_
```

For 8 features accuracy =  0.7695696460402341

```python
fs = SequentialFeatureSelector(lr, k_features=7, forward=True, cv=None)
fs.fit(x,y)
```

```python
fs.k_feature_names_
```

```python
fs.k_score_
```

For 7 features, accuraccy = 0.77734375

We do this recursively for all number of features and then we compare all accuraccy

---

### For Backward Elimination

```python
# Forward = False
fs = SequentialFeatureSelector(lr, k_features=8, forward=False, cv=None)
```

And then same recursively for all number of features
