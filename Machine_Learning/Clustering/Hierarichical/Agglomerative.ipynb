```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster import hierarchy as sc
from sklearn.cluster import AgglomerativeClustering
```

```python
df = pd.read_csv("../../datasets/Iris_unlabeled.csv")
df.head()
```

```python
sns.pairplot(data=df)
plt.show()
```

### Drawing **Dendrogram** to find number of cluster

```python
plt.figure(figsize=(20,20))
sc.dendrogram(sc.linkage(df, method='ward', metric='euclidean')) # single linkage here

plt.savefig('agglomerative_dendogram.png')
plt.show()
```

```python
ac = AgglomerativeClustering(n_clusters=2, linkage='single')
ac.fit_predict(df)
```

```python
df['predict'] = ac.fit_predict(df)
df.head(3)
```

```python
sns.pairplot(data=df, hue='predict')
plt.show()
```
