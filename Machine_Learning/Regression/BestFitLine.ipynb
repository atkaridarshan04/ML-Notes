# Finding the Best Fit Line for Regression

## 1. Understanding the Regression Line

In linear regression, we aim to find the best fit line:

$$ y = mx + b $$

where:
- **y** is the dependent variable (target)
- **x** is the independent variable (feature)
- **m** is the slope (weight/parameter)
- **b** is the intercept (bias term)

---

## 2. Cost Function (Mean Squared Error - MSE)

To measure the error of our model, we use the Mean Squared Error (MSE) as the cost function:

$$ J(m, b) = \frac{1}{2n} \sum_{i=1}^{n} (y_i - (m x_i + b))^2 $$

where:
- **n** is the number of data points
- **y_i** is the actual value
- **(m x_i + b)** is the predicted value

---

## 3. Gradient Descent Algorithm

Gradient Descent is an optimization algorithm used to minimize the cost function iteratively. It updates the parameters as follows:

$$ m := m - \alpha \frac{\partial J}{\partial m} $$
$$ b := b - \alpha \frac{\partial J}{\partial b} $$

where:
  $$
  \frac{\partial J}{\partial m} = -\frac{2}{n} \sum_{i=1}^{n} x_i \left(y_i - (m x_i + b)\right)
  $$

  $$
  \frac{\partial J}{\partial b} = -\frac{2}{n} \sum_{i=1}^{n} \left(y_i - (m x_i + b)\right)
  $$


---

## 4. Python Code to Visualize Gradient Descent

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.linspace(0, 10, 100)
y = 2.5 * x + 5 + np.random.randn(100) * 2  # Linear relation with noise

# Initialize parameters
m, b = 0, 0  # Start with zero slope and intercept
alpha = 0.01  # Learning rate
iterations = 100
m_history, b_history = [], []

# Gradient Descent
for _ in range(iterations):
    y_pred = m * x + b
    error = y - y_pred
    m -= alpha * (-2 / len(x)) * np.sum(x * error)
    b -= alpha * (-2 / len(x)) * np.sum(error)
    m_history.append(m)
    b_history.append(b)

# Plot Gradient Descent Progress
plt.figure(figsize=(10, 6))
for i in range(0, iterations, 10):
    plt.plot(x, m_history[i] * x + b_history[i], linestyle='dashed', label=f'Iteration {i}')
plt.scatter(x, y, color='red', label='Data')
plt.plot(x, m * x + b, color='black', linewidth=2, label='Final Best Fit')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent Optimization')
plt.show()
```

---

## 6. Conclusion

- **Gradient Descent** finds the best `m` and `b` by minimizing the cost function.
- **Learning Rate (\alpha)** affects convergence speed.
- **Stopping Criterion:** When changes in `m` and `b` become small.

By following these steps, we can determine the best-fit regression line for any dataset!
