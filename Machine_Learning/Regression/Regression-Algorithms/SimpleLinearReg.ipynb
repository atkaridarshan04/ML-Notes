# Simple Linear Regression

Simple Linear Regression is a type of Regression algorithms that models the relationship between a dependent variable and a single independent variable.

    y = mx + c

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('../datasets/placement.csv')
df.head(3)
```

```python
df.isnull().sum()
```

```python
x = df[['cgpa']]
y = df['package']

x.ndim, y.ndim
```

```python
# Checking where our data follows linearity

sns.scatterplot(x='cgpa', y='package', data=df)
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
lr = LinearRegression()

lr.fit(x_train, y_train) # This returns the values of 'm' and 'c'
```

Predicting Values

```python
# Checking Accuraccy Score
lr.score(x_test,y_test)
```

```python
#   cgpa	package
#	6.89	3.26

lr.predict([[6.89]])
```

```python
# y = (m * x) + c
# m = coefficient
# c = intercept

print(lr.coef_)
print(lr.intercept_)
```

```python
# There the equation becomes
# y = (0.57425647 * x ) - 1.0270069374542108

y = (0.57425647 * 6.89 ) - 1.0270069374542108
y
```

```python
y_predict = lr.predict(df[['cgpa']])
```

```python
sns.scatterplot(x='cgpa', y='package', data=df)

# Plot the prediction line
plt.plot(df['cgpa'], lr.predict(df[['cgpa']]), c="red")   

plt.legend(["orginal data", "Predicted Line"])

plt.show()
```
