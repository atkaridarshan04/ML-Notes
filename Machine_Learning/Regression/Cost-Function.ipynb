# Cost Function

- A cost function is an important parameter that determines how well a machine
learning model performs for a given dataset.
- Cost function is a measure of how wrong the model is in estimating the relationship
between X(input) and Y(output) Parameter.
- It's also known as a loss function. 

### How it works 
- The cost function calculates the error between the predicted and actual values.
- The cost function is presented as a single number.
- The goal is to minimize the cost function value.
- Lower cost function values indicate better model accuracy.

## 1.Mean Squared Error:
- Mean Squared Error(MSE) is the mean squared difference
between the actual and predicted values. MSE penalizes
high errors caused by outliers by squaring the errors.
- Mean squared error is also known as L2 Loss.



**Advantages:**
- Smooth & Differentiable → MSE has a continuous and differentiable nature, making it ideal for gradient-based optimization like in neural networks.
- Penalizes Large Errors → Squaring the errors gives more weight to larger errors, ensuring the model focuses on big mistakes.

**Disadvantage:**
- Sensitive to Outliers → Since errors are squared, outliers (large errors) can dominate the cost, leading to biased models.
- Doesn't Reflect Absolute Errors Well → Squaring the errors makes it difficult to interpret the actual deviation from the target.

---

## 2.Mean Absolute Error
• Mean Absolute Error(MAE) is the mean absolute difference
between the actual values and the predicted values.
• MAE is more robust to outliers. The insensitivity to outliers is
because it does not penalize high errors caused by outliers.



**Advantages:**
- Robust to Outliers → Unlike MSE, MAE treats all errors equally by taking the absolute difference, reducing sensitivity to outliers.
- Easy to Interpret → Directly represents the average magnitude of errors, making it more intuitive than MSE.

**Disadvantages:**
- Not Smooth for Optimization → The absolute function is not differentiable at zero, which can make gradient-based optimization harder.

---

## 3. Root Mean Square Error
- Root Mean Squared Error (RMSE) is the root squared mean
of the difference between actual and predicted values.
- RMSE can be used in situations where we want to penalize
high errors but not as much as MSE does.



It is differential also and also gives the actual error deviation form the target
