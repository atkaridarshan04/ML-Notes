# Confusion Matrix

A table that summarizes the performance of a classification model by comparing its predicted labels to the true labels



#### Accuracy = TP+TN/N  

#### Error = FN+FP/N

- False Negative: The model has predicted no, but the actual value was
Yes, it is also called as Type-II error.
- False Positive: The model has predicted Yes, but the actual value was
No. It is also called a Type-I error. 

---

#### **We should make sure that all the three values , Precicion Score , Recall and F1 Score should be as high as posssible**

### Precision 
It helps us to measure the ability to classify positive samples in the model.

    Precision : TP/TP +FP

### Recall
It helps us to measure how many positive samples were correctly classified by the ML model.

    Recall: TP/TP+FN

## F1 Score:
It is the harmonic mean of precision and recall. It takes both false positive and
false negatives into account. Therefore, it performs well on an imbalanced
dataset.

    Fl Score = 2 * (Precision * Recall) / ( Precision + Recall)


---

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
df = pd.read_csv('../datasets/Social_Network_Ads.csv')
df.head()
```

```python
# df.isnull().sum()
```

```python
sns.scatterplot(x=df['Age'], y=df['EstimatedSalary'], data=df, hue=df['Purchased'])
plt.show()
```

```python
x = df.iloc[:, :-1]
y = df['Purchased']

x.ndim, y.ndim
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
lr.score(x_test, y_test) 
```

```python
lr.predict(x_test)
```

```python
# 19  19000	 0
lr.predict([[19	,19000]])
```

### Evaluation

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
confusion_matrix(y_test, lr.predict(x_test))
```

```python
sns.heatmap(confusion_matrix(y_test, lr.predict(x_test)), annot=True)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
precision_score(y_test, lr.predict(x_test))*100
```

```python
recall_score(y_test, lr.predict(x_test))*100
```

```python
f1_score(y_test, lr.predict(x_test))*100
```
