```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC # SVC = Support Vector Classifier
```

```python
df = pd.read_csv('../../datasets/polynomial_classification_dataset.csv')
df.head()
```

```python
sns.scatterplot(x=df['X1'], y=df['X2'], data=df, hue=df['output'])
plt.show()
```

```python
x = df.iloc[:, :-1]
y = df['output']

x.ndim, y.ndim
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=4)
```

```python
# 'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'
svc = SVC(kernel="rbf") # try with differnet kernels and select the best one
svc.fit(x_train, y_train)
```

```python
svc.score(x_test, y_test)
```

```python
svc.score(x_train, y_train)
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=svc)
plt.show()
```
