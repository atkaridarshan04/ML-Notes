```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.svm import SVR # SVR = Support Vector Regressor
```

```python
df = pd.read_csv('../datasets/placement_regression.csv')
df.head()
```

```python
sns.scatterplot(x=df['cgpa'], y=df['package'], data=df)
plt.show()
```

```python
x = df[['cgpa']]
y = df['package']

x.ndim, y.ndim
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=4)
```

```python
# 'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'
svr = SVR(kernel='linear')
svr.fit(x_train, y_train)
```

```python
svr.score(x_test, y_test)
```

```python
svr.score(x_train, y_train)
```

```python
sns.scatterplot(x=df['cgpa'], y=df['package'], data=df)
plt.plot(df['cgpa'], svr.predict(x), color='red')
plt.show()
```

---

## Using diffferent kernel

### Polynomial

```python
svr1 = SVR(kernel='poly')
svr1.fit(x_train, y_train)

svr1.score(x_test, y_test), svr1.score(x_train, y_train)
```

```python
sns.scatterplot(x=df['cgpa'], y=df['package'], data=df)
plt.scatter(df['cgpa'], svr1.predict(x), color='red')
plt.show()
```

### RBF

```python
svr2 = SVR(kernel='rbf')
svr2.fit(x_train, y_train)

svr2.score(x_test, y_test), svr2.score(x_train, y_train)
```

```python
sns.scatterplot(x=df['cgpa'], y=df['package'], data=df)
plt.scatter(df['cgpa'], svr2.predict(x), color='red')
plt.show()
```

### Sigmoid

```python
svr3 = SVR(kernel='sigmoid')
svr3.fit(x_train, y_train)

svr3.score(x_test, y_test), svr3.score(x_train, y_train)
```

```python
sns.scatterplot(x=df['cgpa'], y=df['package'], data=df)
plt.scatter(df['cgpa'], svr3.predict(x), color='red')
plt.show()
```
