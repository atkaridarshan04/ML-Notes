```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.datasets import make_moons

from sklearn.svm import SVC
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier
```

```python
x_ ,y_  = make_moons(n_samples=1000, noise=0.2)
```

```python
d = {
    "x1": x_[:, 0],
    "x2": x_[:, 1],
    "y": y_,
}
```

```python
df = pd.DataFrame(d)
df.head(3)
```

```python
sns.scatterplot(x=df['x1'], y=df['x2'], hue=df['y'])
plt.show()
```

```python
x = df.iloc[:, :-1]
y = df['y']

x.ndim, y.ndim
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## SVM

```python
sv = SVC()
sv.fit(x_train, y_train)
```

```python
sv.score(x_test, y_test), sv.score(x_train, y_train)
```

## Bagging 

```python
bg =BaggingClassifier(SVC(), n_estimators=30)
bg.fit(x_train, y_train)
```

```python
bg.score(x_test, y_test), bg.score(x_train, y_train)
```

## Random Forest

```python
rf = RandomForestClassifier(n_estimators=40)
rf.fit(x_train, y_train)
```

```python
rf.score(x_test, y_test), rf.score(x_train, y_train)
```
